```python
import sqlite3
import pandas as pd
```

```python
# PUNTO 2 -- Consulta sobre productos y sus precios en orden ascendente alfabeticamente

SELECT producto, precio_unitario
FROM ventas_tecnologia
ORDER BY producto ASC;
```

```python
# PUNTO 3 -- Ventas con valor mayor a 500 que se realizaron en pais especifico

SELECT *
FROM ventas_tecnologia
WHERE pais = 'Colombia' AND precio_unitario > 500;
```

```python
# PUNTO 4 -- Identificacion de nulos para categoria
SELECT *
FROM ventas_tecnologia
WHERE categoria IS NULL;
```

```python
#PUNTO 5 -- Cuánto dinero generó la empresa agrupado por cada categoría de producto
SELECT categoria, SUM(cantidad * precio_unitario) as ingresos_totales
FROM ventas_tecnologia
GROUP BY categoria;
```

```python
# PUNTO 6 -- Categorias destacadas por su ingreso total
SELECT categoria, SUM(cantidad * precio_unitario) as ingresos_totales
FROM ventas_tecnologia
GROUP BY categoria
HAVING SUM ingresos_totales > 10000;
```
